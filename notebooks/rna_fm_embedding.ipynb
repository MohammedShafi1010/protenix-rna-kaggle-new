```python
import torch, numpy as np, fm
model, alphabet = fm.pretrained.rna_fm_t12()
 
seq = "GGGAAACCCCCC"                      # your ncRNA
token = torch.tensor([alphabet.get_idx(s) for s in seq])[None]
results = model(token, repr_layers=[12])          # 12 → top layer
emb = results["representations"][12].squeeze(0).cpu()
emb=emb.detach().numpy()
emb.shape
 
```
